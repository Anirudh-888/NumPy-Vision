# Extracting RGB Channels from the Real Image

Before applying image filters, I examined the individual color channels of the motorcycle image.

The original image contains three color channels representing red, green, and blue information. By extracting each channel separately, I can observe how different color intensities are distributed throughout the image.

This helps me understand the underlying color information before applying filters that modify or combine the RGB channels.

````python
import numpy as np
import matplotlib.pyplot as plt
image_path="../Images/Input/motorcycle_original.png"
image=plt.imread(image_path)
print("The original image")
plt.imshow(image)
plt.show()
red_channel=image[:,:,0]
green_channel=image[:,:,1]
blue_channel=image[:,:,2]
print("The red channel intensity")
plt.imshow(red_channel, cmap="gray")
plt.show()
print("The green channel intensity")
plt.imshow(green_channel, cmap="gray")
plt.show()
print("The blue channel intensity")
plt.imshow(blue_channel, cmap="gray")
plt.show()
````

# Converting the Real Image to Grayscale

In this section, I convert the motorcycle image from an RGB representation into a grayscale representation.

The original image contains separate red, green, and blue intensity values for every pixel. To create a grayscale image, these color values are combined to produce a single intensity value for each pixel.

Using NumPy's vectorized operations, this conversion can be applied to the entire image at once without using loops. The resulting grayscale image represents variations in brightness rather than color.

````python
import numpy as np 
import matplotlib.pyplot as plt
image=plt.imread(image_path)
red_channel=image[:,:,0]
green_channel=image[:,:,1]
blue_channel=image[:,:,2]
grayscale_image=(red_channel+green_channel+blue_channel)/3
print("The image converted into grey scale")
plt.imshow(grayscale_image,cmap="gray")
plt.show()
````

# Threshold Filtering Using a Boolean Mask

In this section, I apply threshold filtering to the grayscale motorcycle image using a Boolean mask.

A threshold is a numerical value used to separate pixels based on their intensity. By comparing every pixel in the grayscale image with a selected threshold, NumPy creates a Boolean array containing `True` and `False` values.

Pixels that satisfy the condition are represented as `True`, while the remaining pixels are represented as `False`. This demonstrates how Boolean masks can be used to filter and analyze image data using vectorized operations without processing individual pixels manually.

````python
import numpy as np
import matplotlib.pyplot as plt
image=plt.imread("../Images/Input/motorcycle_original.png")
red_channel=image[:,:,0]
green_channel=image[:,:,1]
blue_channel=image[:,:,2]
grayscale_image=(red_channel+green_channel+blue_channel)/3
threshold_mask=grayscale_image>0.5
plt.imshow(threshold_mask,cmap="gray")
plt.show()
````

# Creating a Binary Black-and-White Image

In this section, I convert the Boolean threshold mask into a binary black-and-white representation of the motorcycle image.

The threshold mask separates pixels into two groups based on their grayscale intensity. Pixels that satisfy the threshold condition are represented as `True`, while the remaining pixels are represented as `False`.

By using this Boolean information, I can create a binary image where the selected pixels are represented as white and the remaining pixels are represented as black. This demonstrates how Boolean masks can be used to simplify and segment image data.

````python
import numpy as np
import matplotlib.pyplot as plt
image=plt.imread("../Images/Input/motorcycle_original.png")
red_channel=image[:,:,0]
green_channel=image[:,:,1]
blue_channel=image[:,:,2]
grayscale_image=(red_channel+green_channel+blue_channel)/3
threshold_mask=grayscale_image>0.5
binary_image=np.where(threshold_mask,1,0)
print("The binary image")
plt.imshow(binary_image,cmap='gray')
plt.show()
````

# Exploring Different Threshold Values

In this section, I experiment with different threshold values to observe how they affect the binary representation of the motorcycle image.

The threshold determines which pixels are selected based on their grayscale intensity. Lower threshold values allow more pixels to satisfy the condition, while higher threshold values select fewer pixels.

By comparing different threshold values, I can observe how numerical conditions and Boolean masks affect image segmentation.

````python
import numpy as np
import matplotlib.pyplot as plt
image_path="../Images/Input/motorcycle_original.png"
image=plt.imread(image_path)
red_channel=image[:,:,0]
green_channel=image[:,:,1]
blue_channel=image[:,:,2]
grayscale_image=(red_channel+green_channel+blue_channel)/3
threshold_mask1=grayscale_image>0.3
print("The output image with threshold value as 0.3")
plt.imshow(threshold_mask1, cmap="gray")
plt.show()
threshold_mask2=grayscale_image>0.5
print("The output image with the threshold value as 0.5")
plt.imshow(threshold_mask2,cmap="gray")
plt.show()
threshold_mask3=grayscale_image>0.7
print("The output image with the threshold value as 0.7")
plt.imshow(threshold_mask3,cmap="gray")
plt.show()


````

# Applying Boolean Masks Directly to an RGB Image

A Boolean mask can be used not only to visualize selected pixels but also to directly select or modify pixels in an RGB image.

A Boolean mask is a two-dimensional array containing `True` and `False` values.

For example:

```text
True   → Pixel satisfies the condition
False  → Pixel does not satisfy the condition

````python
import numpy as np
import matplotlib.pyplot as plt
image=plt.imread("../Images/Input/motorcycle_original.png")
copy_image=np.copy(image)
red_channel=image[:,:,0]
green_channel=image[:,:,1]
blue_channel=image[:,:,2]
grayscale_image=(red_channel+green_channel+blue_channel)/3
threshold_mask=grayscale_image>0.5
copy_image[threshold_mask]=[1,0,0]
print("Resultant image after applying the mask to the original image")
plt.imshow(copy_image)
plt.show()
# If the intensity of the grayscale_image in that pixel is more than that of the threshold value then the pixel is changed into red
````
