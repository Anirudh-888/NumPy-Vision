## Modifying a Pixel

A NumPy array can be modified directly using **indexing**.

For an RGB image, the first two dimensions identify the pixel's location:

```text
(row, column)

````python
import numpy as np
rgb_image=np.array([
    [
        [255,0,0],
        [0,255,0],
        [0,0,255]        
    ],
    [
        [255,255,255],
        [0,0,0],
        [255,0,0]
    ],
    [
        [0,255,0],
        [0,0,255],
        [0,0,0]        
    ]
])
print("The color in the location before",rgb_image[1,1])
rgb_image[1,1]=[255,255,0]
print("The color in the location after changing",rgb_image[1,1])
````

# Project Task: Image Cropping

## What I Built

In this step, I built a basic **image cropping operation** using NumPy array slicing.

I extracted a smaller rectangular region from my original RGB image while preserving the RGB values of each selected pixel.

## Concept I Used

### NumPy Array Slicing

I used **NumPy array slicing** to select a range of rows and columns from my multidimensional image array.

The general slicing syntax is:

```text
start : stop

## Cropped the image as 2*2

````python
import numpy as np
rgb_image=np.array([
    [
        [255,0,0],
        [0,255,0],
        [0,0,255]        
    ],
    [
        [255,255,255],
        [0,0,0],
        [255,0,0]
    ],
    [
        [0,255,0],
        [0,0,255],
        [0,0,0]        
    ]
])
print("The original image\n",rgb_image)
cropped_image=rgb_image[0:2,0:2,0:3]
print("The cropped image is\n",cropped_image)
display(np.shape(cropped_image))
display(cropped_image.ndim)
display(np.size(cropped_image))
````
