# Accessing Regions of an Image Using NumPy Slicing

After exploring individual pixels and RGB channel values, I extended the same indexing concept to access larger regions of the image.

Since an RGB image is represented as a three-dimensional NumPy array with the structure:

`(height, width, channels)`

I can use NumPy slicing to select a rectangular region of the image.

## Key Concept

A region of an image can be accessed using:

`image[start_row:end_row, start_column:end_column]`

When the channel axis is not specified, NumPy preserves all the RGB channel values for the selected region.

I can also explicitly represent the complete structure as:

`image[start_row:end_row, start_column:end_column, :]`

Here, `:` means that all color channels are included.

## Why This Matters

Image slicing allows me to work with specific regions instead of processing the entire image. This concept is essential for image cropping, region-based pixel manipulation, image preprocessing, and computer vision applications.

# Modifying Individual Color Channels

In this section, I applied NumPy indexing and slicing to manipulate individual RGB channels within a selected region of a real image.

Instead of replacing all three color values of each pixel, I can target a specific channel using the third axis of the image array.

For an RGB image:

# Modifying Pixel Intensity in a Real Image

In this section, I applied mathematical operations directly to the pixel values of a real image.

Instead of replacing pixel values with completely new values, I modified the existing numerical values to observe how mathematical operations affect the visual appearance of an image.

This demonstrates an important connection between NumPy operations and image processing: changes to numerical pixel data directly influence properties such as color intensity and brightness.

To preserve the original image, I performed these operations on a copy of the image.

````python
import numpy as np
import matplotlib.pyplot as plt
image_path="../Images/Input/motorcycle_original.png"
image=plt.imread(image_path)
modified_image=image.copy()
modified_image[0:200,0:200]=modified_image[0:200,0:200]*0.3
plt.imshow(modified_image)
plt.show()
````

# Increasing Pixel Intensity in a Real Image

In this section, I increased the intensity of pixel values within a selected region of the real image.

By multiplying the existing pixel values by a value greater than `1`, I can increase their intensity. This can make the selected region appear brighter.

While performing this operation, it is important to consider the valid range of pixel values. Since the image is represented using normalized values between `0` and `1`, increasing the values beyond this range may require additional handling.

This demonstrates how mathematical operations on NumPy arrays can be used to control the visual intensity of image data.

# Controlling Pixel Values with NumPy Clip

In this section, I applied `np.clip()` to control the range of pixel values after modifying the intensity of a real image.

When pixel values are increased through mathematical operations, some values can exceed the valid range of the image. Since the motorcycle image uses normalized pixel values between `0` and `1`, I used `np.clip()` to ensure that the modified values remain within this range.

This demonstrates how NumPy can be used not only to manipulate image data but also to control and maintain valid numerical values during image processing.

````python
import numpy as np
import matplotlib.pyplot as plt
image=plt.imread(image_path)
edited_image=image.copy()
edited_image[0:200,0:200]=edited_image[0:200,0:200]*2
edited_image[0:200,0:200]=np.clip(edited_image[0:200,0:200],0,1)
print("The image after increasing the intensity at a region")
plt.imshow(edited_image)
plt.show() 
````

# Modifying the Brightness of the Entire Image

In this section, I extended the pixel intensity manipulation from a selected region to the entire image.

Since the complete image is represented as a NumPy array, mathematical operations can be applied directly to all pixel values at once. By modifying the numerical intensity of every pixel, I can change the overall brightness of the image.

When increasing pixel values, it is important to ensure that they remain within the valid range of the image. I use `np.clip()` to constrain the resulting values between `0` and `1`.

This demonstrates how NumPy operations can be applied efficiently to an entire image without manually processing individual pixels.

````python
import numpy as np
import matplotlib.pyplot as plt
image=plt.imread(image_path)
print("The original image")
plt.imshow(image)
plt.show()
edited_image=image.copy()
edited_image=np.clip(edited_image*2,0,1)
print("Image with increased brightness")
plt.imshow(edited_image)
plt.show()
````

# Decreasing the Brightness of the Entire Image

In this section, I decreased the overall brightness of the real image by applying a mathematical operation to all of its pixel values.

Since the complete image is represented as a NumPy array, multiplying the entire array by a positive value between `0` and `1` reduces the intensity of every pixel.

For example:

- Multiplying by `0.5` reduces the pixel intensity by half.
- Multiplying by `0.25` reduces the pixel intensity further.

Since the image contains normalized pixel values between `0` and `1`, multiplying them by a positive value between `0` and `1` does not cause the values to exceed the valid maximum range.

Therefore, `np.clip()` is not required for this operation.

This demonstrates how NumPy can efficiently modify the intensity of every pixel in an image using a single element-wise mathematical operation.

````python
import numpy as np
import matplotlib.pyplot as plt
image=plt.imread(image_path)
print("The original image")
plt.imshow(image)
plt.show()
edited_image=image.copy()
edited_image=edited_image*0.5
print("Image with decreased brightness")
plt.imshow(edited_image)
plt.show()
````
