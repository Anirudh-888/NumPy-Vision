# Step 7: Image Transformations

## What I am doing

In this step, I am applying **image transformations** to manipulate the position and orientation of pixels in my image.

## Concept I am using

I am using **NumPy array manipulation** to transform an image by changing the arrangement of its rows and columns while keeping the RGB values of each pixel intact.

### Transformations I will learn

- **Flipping** — reversing the order of pixels along an axis.
- **Rotating** — changing the orientation of the image by rearranging its dimensions.
- **Transposing** — interchanging the rows and columns of an array.

## What I am learning

I am learning how operations on a NumPy array directly affect the spatial arrangement of pixels in an image.

This helps me understand how NumPy can be used for practical image processing and forms a foundation for later computer vision operations.

### Flipping the image

```python
import numpy as np
import matplotlib.pyplot as plt
rgb_image=np.array([
    [
        [255,0,0],
        [0,255,0],
        [0,0,255]        
    ],
    [
        [255,255,255],
        [0,0,0],
        [255,0,0]
    ],
    [
        [0,255,0],
        [0,0,255],
        [0,0,0]        
    ]
])
print("The original image is")
display(plt.imshow(rgb_image))
plt.show()
print("Flipping the image")
flipped_image=rgb_image[:,::-1,:]
plt.imshow(flipped_image)
plt.show()
```

# Vertical Image Flipping

## What I am doing

In this step, I am performing a **vertical flip** on an RGB image using NumPy array slicing.

## Concept I am using

An RGB image is represented as a three-dimensional NumPy array with the structure:

`(rows, columns, RGB channels)`

To flip an image vertically, I need to reverse the order of the **rows** while keeping the columns and RGB channels unchanged.

## What changes

- Rows → Reversed
- Columns → Unchanged
- RGB channels → Unchanged

## What I learned

I learned that image transformations can be performed by manipulating specific axes of a NumPy array. By reversing the row axis, I can flip an image vertically without changing its dimensions or pixel color values.

```python
import numpy as np
import matplotlib.pyplot as plt
rgb_image=np.array([
    [
        [255,0,0],
        [0,255,0],
        [0,0,255]        
    ],
    [
        [255,255,255],
        [0,0,0],
        [255,0,0]
    ],
    [
        [0,255,0],
        [0,0,255],
        [0,0,0]        
    ]
])
print("The original image is")
plt.imshow(rgb_image)
plt.show()
print("Performing vertical flip")
vertically_Flipped_Image=rgb_image[::-1,:,:]
plt.imshow(vertically_Flipped_Image)
plt.show()
```

# 90° Clockwise Image Rotation

## What I am doing

In this step, I am rotating an RGB image by **90 degrees clockwise** using NumPy array transformations.

## Concepts I am using

I am combining my understanding of **transpose** and **flipping** to rearrange the spatial positions of pixels in the image.

A rotation changes the positions of rows and columns while preserving the RGB values of each pixel.

## How I am thinking about rotation

Instead of manually moving every pixel to a new position, I can perform image rotation by combining array operations such as transposing and flipping.

## Why this is important

Understanding image rotation helps me learn how NumPy can manipulate image data efficiently. These transformations are also useful in computer vision and machine learning for image preprocessing and data augmentation.

```python
import numpy as np
import matplotlib.pyplot as plt
rgb_image=np.array([
    [
        [255,0,0],
        [0,255,0],
        [0,0,255]        
    ],
    [
        [255,255,255],
        [0,0,0],
        [255,0,0]
    ],
    [
        [0,255,0],
        [0,0,255],
        [0,0,0]        
    ]
])
print("The original image")
plt.imshow(rgb_image)
plt.show()
transposed_image=np.transpose(rgb_image,axes=(1,0,2))
plt.imshow(transposed_image)
plt.show()
flipped_image=np.flip(transposed_image,axis=1)
print("The completely rotated image is")
plt.imshow(flipped_image)
plt.show()

```

# 90° Counterclockwise Image Rotation

## What I am doing

In this step, I am rotating an RGB image by **90 degrees counterclockwise** using NumPy array transformations.

## Concepts I am using

I am combining **array transposition** and **array flipping** to rearrange the positions of pixels in the image.

## How the rotation works

First, I transpose the image to swap its spatial dimensions:

- Rows become columns
- Columns become rows
- RGB channels remain unchanged

After transposing the image, I flip the **rows (axis 0)** to produce a 90° counterclockwise rotation.

## Transformation Process

Original Image  
↓  
Transpose  
↓  
Flip Rows (Axis 0)  
↓  
90° Counterclockwise Rotation

## What I learned

I learned that different image rotations can be created by combining NumPy operations. A 90° clockwise rotation and a 90° counterclockwise rotation both use transposition, but they differ in the axis that is flipped afterward.

```python
import numpy as np 
import matplotlib.pyplot as plt
rgb_image=np.array([
    [
        [255,0,0],
        [0,255,0],
        [0,0,255]        
    ],
    [
        [255,255,255],
        [0,0,0],
        [255,0,0]
    ],
    [
        [0,255,0],
        [0,0,255],
        [0,0,0]        
    ]
])
print("The orginal image is")
plt.imshow(rgb_image)
plt.show()
transposed_image=np.transpose(rgb_image, axes=(1,0,2))
counter_clock_wise_flipped_image=np.flip(transposed_image,axis=(0))
print("The counter clock wise flipped image")
plt.imshow(counter_clock_wise_flipped_image)
plt.show()
```

# 180° Image Rotation

## What I am doing

In this step, I am rotating an RGB image by **180 degrees** using NumPy array manipulation.

## Concepts I am using

I am using my understanding of array axes and flipping to rearrange the positions of pixels in the image.

A 180-degree rotation reverses the image both vertically and horizontally.

## How the rotation works

To rotate an image by 180 degrees, I need to reverse:

- The rows
- The columns

The RGB channels remain unchanged because the color information of each pixel should not be modified.

## Transformation Process

Original Image  
↓  
Reverse Rows (Axis 0)  
↓  
Reverse Columns (Axis 1)  
↓  
180° Rotated Image

## What I learned

I learned that a complex-looking image transformation can be performed by manipulating the appropriate axes of a NumPy array. By reversing both spatial axes, I can rotate an image by 180 degrees while preserving its RGB color information.

```python
import numpy as np
import matplotlib.pyplot as plt
rgb_image=np.array([
    [
        [255,0,0],
        [0,255,0],
        [0,0,255]        
    ],
    [
        [255,255,255],
        [0,0,0],
        [255,0,0]
    ],
    [
        [0,255,0],
        [0,0,255],
        [0,0,0]        
    ]
])
print("The original image is")
plt.imshow(rgb_image)
plt.show()
image_180_flipped=np.flip(rgb_image,axis=(0,1))
print("The 180 flipped image is")
plt.imshow(image_180_flipped)
plt.show()
```
