# Creating a 3D NumPy Array

A NumPy array can have multiple dimensions. A **3D array** contains data organized across three dimensions.

For an RGB image, a 3D array can represent the image using the structure:

$$
(\text{height},\ \text{width},\ \text{channels})
$$

For example:

$$
(3,\ 3,\ 3)
$$

This represents:

| Dimension | Meaning | Value |
|:---|:---|:---:|
| 1st | Height / Number of rows | 3 |
| 2nd | Width / Pixels per row | 3 |
| 3rd | RGB channels per pixel | 3 |

Each pixel contains three values representing its **Red, Green, and Blue** channels:

```text
[R, G, B]

````python
import numpy as np
image=np.array([
    [
        ["R","G","B"],
        ["R","G","B"],
        ["R","G","B"]
    ],
    [
        ["R","G","B"],
        ["R","G","B"],
        ["R","G","B"]
    ],
    [
        ["R","G","B"],
        ["R","G","B"],
        ["R","G","B"]
    ]
])
print(image)
np.shape(image)
````

# Representing RGB Pixel Values


````python
import numpy as np
rgb_image=np.array([
    [
        [255,0,0],
        [0,255,0],
        [0,0,255]        
    ],
    [
        [255,255,255],
        [0,0,0],
        [255,0,0]
    ],
    [
        [0,255,0],
        [0,0,255],
        [0,0,0]        
    ]
])
np.shape(rgb_image)
rgb_image.ndim
np.size(rgb_image)
rgb_image.dtype
print("The blue pixel from row 0, column 2",rgb_image[0,2])
print("The white pixel from row 1, column 1",rgb_image[1,1])
````

## Modifying a Pixel

A NumPy array can be modified directly using **indexing**.

For an RGB image, the first two dimensions identify the pixel's location:

```text
(row, column)

````python
import numpy as np
rgb_image=np.array([
    [
        [255,0,0],
        [0,255,0],
        [0,0,255]        
    ],
    [
        [255,255,255],
        [0,0,0],
        [255,0,0]
    ],
    [
        [0,255,0],
        [0,0,255],
        [0,0,0]        
    ]
])
print("The color in the location before",rgb_image[1,1])
rgb_image[1,1]=[255,255,0]
print("The color in the location after changing",rgb_image[1,1])
````
