# Creating a 3D NumPy Array

A NumPy array can have multiple dimensions. A **3D array** contains data organized across three dimensions.

For an RGB image, a 3D array can represent the image using the structure:

$$
(\text{height},\ \text{width},\ \text{channels})
$$

For example:

$$
(3,\ 3,\ 3)
$$

This represents:

| Dimension | Meaning | Value |
|:---|:---|:---:|
| 1st | Height / Number of rows | 3 |
| 2nd | Width / Pixels per row | 3 |
| 3rd | RGB channels per pixel | 3 |

Each pixel contains three values representing its **Red, Green, and Blue** channels:

```text
[R, G, B]

````python
import numpy as np
image=np.array([
    [
        ["R","G","B"],
        ["R","G","B"],
        ["R","G","B"]
    ],
    [
        ["R","G","B"],
        ["R","G","B"],
        ["R","G","B"]
    ],
    [
        ["R","G","B"],
        ["R","G","B"],
        ["R","G","B"]
    ]
])
print(image)
np.shape(image)
````

# Representing RGB Pixel Values


````python
import numpy as np
rgb_image=np.array([
    [
        [255,0,0],
        [0,255,0],
        [0,0,255]        
    ],
    [
        [255,255,255],
        [0,0,0],
        [255,0,0]
    ],
    [
        [0,255,0],
        [0,0,255],
        [0,0,0]        
    ]
])
display(np.shape(rgb_image))
display(rgb_image.ndim)
display(np.size(rgb_image))
display(rgb_image.dtype)
print("The blue pixel from row 0, column 2",rgb_image[0,2])
print("The white pixel from row 1, column 1",rgb_image[1,1])
````

# Accessing Individual RGB Channels

In this section, I explored how individual color channels are represented within a pixel of an RGB image. By treating the image as a three-dimensional NumPy array, I was able to access specific color values using row, column, and channel indices.

## Key Concept

An RGB image is structured as:

`(height, width, channels)`

Each pixel contains three color channel values:

- Channel 0 — Red
- Channel 1 — Green
- Channel 2 — Blue

Using NumPy indexing, I can access a specific color value using:

`image[row, column, channel]`

Understanding this structure is important because it provides the foundation for pixel-level manipulation, image processing, and computer vision applications.

````python
import numpy as np
import matplotlib.pyplot as plt 
image=plt.imread("../Images/Input/motorcycle_original.png")
red,green,blue=image[100,100]
print("Accessing the values through tuple unpacking")
print(f"The value of red in the pixel is:{red}")
print(f"The value of green in the pixel is:{green}")
print(f"The value of blue in the pixel is:{blue}")
print("\nAccessing the values individually")
red=image[100,100,0]
print(f"The value of red in the pixel is:{red}")
green=image[100,100,1]
print(f"The value of green in the pixel is:{green}")
blue=image[100,100,2]
print(f"The value of blue in the pixel is:{blue}")
````

# Accessing Regions of an Image Using NumPy Slicing

After exploring individual pixels and RGB channel values, I extended the same indexing concept to access larger regions of the image.

Since an RGB image is represented as a three-dimensional NumPy array with the structure:

`(height, width, channels)`

I can use NumPy slicing to select a rectangular region of the image.

## Key Concept

A region of an image can be accessed using:

`image[start_row:end_row, start_column:end_column]`

When the channel axis is not specified, NumPy preserves all the RGB channel values for the selected region.

I can also explicitly represent the complete structure as:

`image[start_row:end_row, start_column:end_column, :]`

Here, `:` means that all color channels are included.

## Why This Matters

Image slicing allows me to work with specific regions instead of processing the entire image. This concept is essential for image cropping, region-based pixel manipulation, image preprocessing, and computer vision applications.
